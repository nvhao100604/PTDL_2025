## Công thức tính độ lệch chuẩn

Công thức tính độ lệch chuẩn được thể hiện dưới dạng công thức:

$$
s = \sqrt{\frac{\sum (x_i - \bar{x})^2}{n - 1}}
$$

Trong đó:

- **S** là độ lệch tiêu chuẩn  
- **xᵢ** là kết quả quan sát thứ i của mẫu  
- **x̄** là giá trị trung bình của các quan sát này  
- **n** là số lượng quan sát trong mẫu.


```python
std_per_movie = data.groupby("title")["rating"].std().dropna().sort_values(ascending=False)
top_std_movies = std_per_movie.head(10)

print("\n🎬 Top 10 phim có độ lệch chuẩn rating cao nhất:")
print(top_std_movies)
```

```python
# Vẽ biểu đồ Top 10 phim gây tranh cãi
plt.figure(figsize=(10,5))
top_std_movies.sort_values().plot(kind='barh', color='orange')
plt.title("Top 10 phim có độ phân hóa ý kiến mạnh nhất (Độ lệch chuẩn rating cao nhất)")
plt.xlabel("Độ lệch chuẩn (std)")
plt.ylabel("Tên phim")
plt.savefig("images/top10_std_movies.png", dpi=300, bbox_inches='tight')
plt.show()
```

## 🎯 Ý nghĩa công thức

$$
std_{phim} > \mu_{std} + 2\sigma_{std}
$$

Công thức dùng để **tìm các bộ phim có độ lệch chuẩn cao bất thường** — tức là những phim có ý kiến khán giả rất phân hóa (một số người chấm rất thấp, một số chấm rất cao).

---

## 🧮 Giải thích từng giá trị

| Ký hiệu | Tên gọi | Ý nghĩa | Cách tính |
|----------|----------|----------|------------|
| std_phim | Độ lệch chuẩn của từng phim | Đo mức độ phân tán của điểm đánh giá từng bộ phim | s = sqrt( Σ(xᵢ - x̄)² / (n - 1) ) |
| μ_std | Giá trị trung bình của tất cả độ lệch chuẩn phim | Trung bình độ biến thiên trong toàn bộ tập phim | μ_std = ( Σ std_phim ) / N |
| σ_std | Độ lệch chuẩn của các giá trị std_phim | Đo xem độ lệch chuẩn giữa các phim biến động mạnh hay yếu | σ_std = sqrt( Σ(std_phim - μ_std)² / (N - 1) ) |


---

## ✅ Các bước tính

1. **Tính `std_phim`** cho từng phim  
   → dựa vào cột `rating` của từng phim.

2. **Tính `μ_std`**  
   → trung bình của toàn bộ các `std_phim`.

3. **Tính `σ_std`**  
   → độ lệch chuẩn của danh sách các `std_phim`.

4. **Xác định ngưỡng:**  
   $$
   \text{Ngưỡng} = \mu_{std} + 2\sigma_{std}
   $$

5. **Lọc phim đặc biệt:**  
   Chọn các phim có:
   $$
   std_{phim} > \mu_{std} + 2\sigma_{std}
   $$
   → tức là những phim thuộc **2.5% cao nhất**, thể hiện mức độ phân hóa cực mạnh trong đánh giá khán giả.

---

## 🧩 Ví dụ minh họa

| Phim | std_phim |
|------|-----------|
| A | 0.9 |
| B | 1.1 |
| C | 0.7 |
| D | 2.0 |
| E | 0.8 |

Tính:

$$
\mu_{std} = \frac{0.9 + 1.1 + 0.7 + 2.0 + 0.8}{5} = 1.1
$$

$$
\sigma_{std} = \sqrt{\frac{(0.9-1.1)^2 + (1.1-1.1)^2 + (0.7-1.1)^2 + (2.0-1.1)^2 + (0.8-1.1)^2}{5-1}} = 0.47
$$

$$
$$

(b) So sánh độ lệch chuẩn trung bình giữa các thập niên

```python
# Trích năm phát hành từ tiêu đề
import re
def extract_year(title):
    match = re.search(r'\((\d{4})\)', str(title))
    return int(match.group(1)) if match else np.nan

data["year"] = data["title"].apply(extract_year)

# Nhóm thập niên
def decade(y):
    if pd.isna(y): return np.nan
    return int(y // 10 * 10)

data["decade"] = data["year"].apply(decade)

# Tính độ lệch chuẩn trung bình rating theo thập niên
std_by_decade = data.groupby("decade")["rating"].std()
```

```python
plt.figure(figsize=(8,5))
std_by_decade.plot(kind="bar", color="steelblue")
plt.title("Độ lệch chuẩn rating trung bình theo thập niên sản xuất phim")
plt.xlabel("Thập niên")
plt.ylabel("Độ lệch chuẩn (σ)")

plt.savefig("images/std_by_decade.png", dpi=300, bbox_inches='tight')

plt.show()

print("\n📈 Độ lệch chuẩn trung bình theo thập niên:")
print(std_by_decade)
```

- Trước tiên, cần lọc ra các bộ phim nhận được ít nhất 250 lượt đánh giá (số này mang tính tùy ý).
    - Để làm điều này, tôi nhóm dữ liệu theo cột title và dùng size() để lấy một Series cho biết kích thước của mỗi nhóm (tức số lượt đánh giá của từng phim):

```python
ratings_by_title = data.groupby("title").size()
ratings_by_title.head()
active_titles = ratings_by_title.index[ratings_by_title >= 250]
active_titles
```

- Chỉ số (index) của các bộ phim nhận được ít nhất 250 lượt đánh giá sau đó có thể được sử dụng để chọn các hàng tương ứng từ `mean_ratings` bằng cách dùng `.loc`:

```python
mean_ratings = mean_ratings.loc[active_titles]
mean_ratings
```

```python
mean_ratings = mean_ratings.rename(index={"Seven Samurai (The Magnificent Seven) (Shichinin no samurai) (1954)":
                           "Seven Samurai (Shichinin no samurai) (1954)"})
```

- Để xem các bộ phim được nữ khán giả đánh giá cao nhất, chúng ta có thể sắp xếp theo cột `F` (giới tính nữ) theo thứ tự giảm dần:

```python
top_female_ratings = mean_ratings.sort_values("F", ascending=False)
top_female_ratings.head()
```

## Đo lường sự khác biệt về đánh giá

- Giả sử bạn muốn tìm những bộ phim mà có sự khác biệt lớn về đánh giá giữa nam và nữ.
    - Một cách làm là:
        - Thêm một cột vào `mean_ratings` chứa hiệu số trung bình giữa hai giới tính.
        - Sắp xếp theo cột này để tìm các bộ phim có sự khác biệt lớn nhất:

```python
mean_ratings["diff"] = mean_ratings["M"] - mean_ratings["F"]
```

- Sắp xếp theo cột `diff` sẽ cho ra những bộ phim có sự khác biệt điểm đánh giá lớn nhất, giúp chúng ta xem phim nào được nữ khán giả yêu thích hơn.

```python
sorted_by_diff = mean_ratings.sort_values("diff")
sorted_by_diff.head()
```

- Đảo ngược thứ tự các hàng và lấy 10 hàng đầu một lần nữa, ta sẽ có danh sách các bộ phim được nam khán giả yêu thích nhưng nữ khán giả không đánh giá cao.

```python

sorted_by_diff[::-1].head()
```

- Giả sử khi muốn tìm những bộ phim gây ra nhiều tranh cãi nhất trong số các khán giả, bất kể giới tính.
- Sự khác biệt (disagreement) có thể được đo bằng phương sai (variance) hoặc độ lệch chuẩn (standard deviation) của các lượt đánh giá.
- Để làm điều này, trước tiên ta tính độ lệch chuẩn của điểm đánh giá theo từng phim (title), sau đó lọc ra các phim hoạt động nhiều (có đủ lượt đánh giá):

```python
rating_std_by_title = data.groupby("title")["rating"].std()
rating_std_by_title = rating_std_by_title.loc[active_titles]
rating_std_by_title.head()
```

- Sau đó, chúng ta sắp xếp theo thứ tự giảm dần và chọn 10 hàng đầu, tương đương với 10 bộ phim có sự khác biệt về đánh giá lớn nhất.

```python
rating_std_by_title.sort_values(ascending=False)[:10]
```
