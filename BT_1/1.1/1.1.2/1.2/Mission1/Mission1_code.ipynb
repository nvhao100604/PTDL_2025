# Nhiệm vụ 1

1. Tính mean, median, mode, variance, standard deviation, range, percentile, quartile, interquartile range
(IQR) sử dụng thư viện numpy và stats trên tập dữ liệu COVID.
- Giải thích:
    - mean: Giá trị trung bình cộng của dữ liệu.
    - median: Giá trị nằm giữa khi dữ liệu được sắp xếp.
    - mode: Giá trị xuất hiện nhiều nhất trong dữ liệu.
    - variance: Đo độ phân tán của dữ liệu quanh giá trị trung bình.
    - standard deviation: Độ lệch chuẩn, căn bậc hai của phương sai.
    - range: Khoảng cách giữa giá trị lớn nhất và nhỏ nhất.
    - percentile: Giá trị tại một phần trăm cụ thể của dữ liệu đã sắp xếp.
    - quartile: Các giá trị chia dữ liệu thành bốn phần bằng nhau.
    - interquartile range (IQR): Khoảng giữa tứ phân vị thứ 1 (Q1) và thứ 3 (Q3).


## Tải dữ liệu

```python
import requests
import os

# URL của file CSV
url = "https://catalog.ourworldindata.org/garden/covid/latest/compact/compact.csv"

# Nơi lưu file
save_path = "data/compact.csv"

# Tạo thư mục nếu chưa tồn tại
os.makedirs(os.path.dirname(save_path), exist_ok=True) 

# Tải file từ URL
response = requests.get(url)

# Kiểm tra xem yêu cầu tải có thành công không
if response.status_code == 200:
    # Lưu file vào đường dẫn chỉ định
    with open(save_path, 'wb') as file:
        file.write(response.content)
    print(f"File đã được tải về và lưu tại: {save_path}")
else:
    print(f"Lỗi khi tải file. Mã lỗi: {response.status_code}")
```

## Lọc các cột, biến đổi tên và lưu thành file mới để thao tác

```python
# Chọn các cột cần lọc
selected_columns = [
    'continent',
    'country',
    'date', 
    'total_cases', 
    'new_cases'
]

# Kiểm tra xem các cột có tồn tại trong file không
missing_cols = [col for col in selected_columns if col not in covid_data.columns]
if missing_cols:
    print(f"Các cột không tồn tại trong file: {missing_cols}")

# Lọc các cột
filtered_df = covid_data[selected_columns]

# Đổi tên cột
rename_dict = {
    'continent': 'continent',
    'country': 'location',
    'date': 'date',
    'total_cases': 'total_cases',
    'new_cases': 'new_cases'
}

# Áp dụng đổi tên cột
filtered_df = filtered_df.rename(columns=rename_dict)

# Tạo thư mục đầu ra nếu chưa tồn tại
output_path = "data/covid-data.csv"

os.makedirs(os.path.dirname(output_path), exist_ok=True)

# Lưu file CSV mới
filtered_df.to_csv(output_path, index=False)
print(f"File đã được lọc và lưu tại: {output_path}")

# Hiển thị 5 dòng đầu tiên của file mới
print("\n5 dòng đầu tiên của file mới:")
print(filtered_df.head())
```

## Tiến hành thực hiện như trong nhiệm vụ

```python
import numpy as np
import pandas as pd
from scipy import stats

# Load the .csv into a dataframe using read_csv
covid_data = pd.read_csv("data/covid-data.csv")
covid_data = covid_data[['continent',
'location','date','total_cases','new_cases']]
```

```python
# Kiểm tra dữ liệu
covid_data.info()
covid_data.describe()
# Take a quick look at the data
covid_data.head(5)
covid_data.dtypes
covid_data.shape
```

```python
# Get the mean of the data
# Tính trung bình của dữ liệu 
data_mean = np.mean(covid_data["new_cases"]) 

# Get the median of the data
# Tính trung vị của dữ liệu
data_median = np.median(covid_data["new_cases"])

# Get the mode of the data
# Tính mode của dữ liệu
data_mode = stats.mode(covid_data["new_cases"])

# Obtain the variance of the data
# Tính phương sai của dữ liệu
data_variance = np.var(covid_data["new_cases"])

# Obtain the standard deviation of the data
# Tính độ lệch chuẩn của dữ liệu
data_sd = np.std(covid_data["new_cases"])

# Compute the maximum and minimum values of the data
# Tính giá trị lớn nhất và nhỏ nhất của dữ liệu
data_max = np.max(covid_data["new_cases"])
data_min = np.min(covid_data["new_cases"])

# Obtain the 60th percentile of the data
# Tính phần trăm vị thứ 60 của dữ liệu
data_percentile = np.percentile(covid_data["new_cases"],60)

# Obtain the quartiles of the data
# Tính phần tư của dữ liệu
data_quartile = np.quantile(covid_data["new_cases"],0.75)

# Get the IQR of the data
# Tính khoảng tứ phân vị của dữ liệu
data_IQR = stats.iqr(covid_data["new_cases"])
```

```python
# In kết quả
print("Mean: ", data_mean)
print("Median: ", data_median)
print("Mode: ", data_mode.mode)
print("Variance: ", data_variance)
print("Standard Deviation: ", data_sd)
print("Max: ", data_max)
print("Min: ", data_min)
print("60th Percentile: ", data_percentile)
print("75th Percentile (Quartile): ", data_quartile)
print("IQR: ", data_IQR)
```
