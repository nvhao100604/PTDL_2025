## 1. Import thư viện và nạp dữ liệu vào notebook

```python
import pandas as pd
# Đọc dữ liệu từ file CSV
marketing_data = pd.read_csv("data/marketing_campaign.csv")

# Lọc các cột cần thiết cho phân tích
marketing_data = marketing_data[['ID','Year_Birth', 'Education',
'Marital_Status','Income','Kidhome', 'Teenhome',
'Dt_Customer', 'Recency','NumStorePurchases',
'NumWebVisitsMonth']]
```

**Ý nghĩa các cột**
- ID: Mã định danh duy nhất của khách hàng.
- Year_Birth: Năm sinh của khách hàng.
- Education: Trình độ học vấn của khách hàng (ví dụ: Graduation, PhD, Master, Basic, 2n Cycle).
- Marital_Status: Tình trạng hôn nhân của khách hàng (ví dụ: Married, Single, Together, Divorced).
- Income: Thu nhập hàng năm của hộ gia đình khách hàng.
- Kidhome: Số lượng trẻ nhỏ trong gia đình khách hàng.
- Teenhome: Số lượng thanh thiếu niên (teen) trong gia đình khách hàng.
- Dt_Customer: Ngày khách hàng đăng ký thành khách hàng của công ty.
- Recency: Số ngày kể từ lần mua hàng cuối cùng của khách hàng.
- NumStorePurchases: Số lần mua hàng được thực hiện trực tiếp tại cửa hàng.
- NumWebVisitsMonth: Số lần khách hàng truy cập trang web của công ty trong tháng gần nhất.

```python
# Kiểm tra thông tin cơ bản của dữ liệu
print("Thông tin vê dữ liệu marketing_data:")
marketing_data.info()
# Thống kê mô tả của dữ liệu
print(f"\nThống kê mô tả của dữ liệu marketing_data: \n{marketing_data.describe()}")

# Hiển thị 5 dòng đầu tiên của dữ liệu
print(f"\n5 dòng đầu tiên của dữ liệu marketing_data: \n{marketing_data.head()}")

# Kiểm tra kiểu dữ liệu và kích thước của DataFrame
print(f"\nKiểu dữ liệu của các cột trong marketing_data: \n{marketing_data.dtypes}")

# Kiểm tra kích thước của DataFrame
print(f"\nKích thước của DataFrame marketing_data (số dòng, số cột): {marketing_data.shape}")
```

## 2. Thực hiện các yêu cầu trong nhiệm vụ 2

### Loại bỏ dữ liệu trùng lặp

```python
marketing_data.head()
# Remove duplicates across the columns in our dataset:
marketing_data_duplicate = marketing_data.drop_duplicates()
# Delete a specified row at index value 1:
marketing_data.drop(labels=[1], axis=0)
# Delete a single column
marketing_data.drop(labels=['Year_Birth'], axis=1) 
```

**Ý nghĩa công việc**
- Xem 5 dòng đầu tiên:
    - Cho phép xem nhanh cấu trúc, định dạng và một vài mẫu dữ liệu ban đầu. Đây là bước kiểm tra chất lượng dữ liệu cơ bản.
- Xóa các dòng trùng lặp: 
    - Tạo ra một tập dữ liệu mới (marketing_data_duplicate) chỉ chứa các hàng duy nhất. Điều này giúp đảm bảo mỗi quan sát (khách hàng) chỉ được tính một lần, tránh làm sai lệch kết quả phân tích thống kê.
- Xóa một dòng cụ thể (tại chỉ mục 1): 
    - Loại bỏ một hàng dữ liệu cụ thể, có thể là do hàng đó chứa lỗi nghiêm trọng (ví dụ: giá trị thiếu quá nhiều, dữ liệu không hợp lệ) mà không thể sửa chữa. axis=0 chỉ định rằng bạn đang làm việc với các hàng (index).
- Xóa một cột cụ thể (cột 'Year_Birth'): 
    - Xóa cột có tên 'Year_Birth' khỏi DataFrame marketing_data, trả về một DataFrame mới (bản đã xóa cột). Bản gốc marketing_data KHÔNG thay đổi nếu không gán kết quả hoặc không dùng inplace=True.

### Thay thế dữ liệu và thay đổi định dạng của dữ liệu

```python
# Replace the values in Teenhome with has teen and has no teen
marketing_data['Teenhome_replaced'] = marketing_data['Teenhome'].replace([0,1,2],['has no teen','has teen','has teen'])
# Fill NAs in the Income column
marketing_data['Income'] = marketing_data['Income'].fillna(0)
# Change the data type of the Income column from float to int
marketing_data['Income_changed'] = marketing_data['Income'].astype(int)
```

**Ý nghĩa công việc**
- Chuyển đổi biến định lượng thành biến phân loại:
    - Chuyển đổi số lượng thanh thiếu niên (0, 1, 2) thành một biến phân loại đơn giản hơn: 'has no teen' (không có teen) hoặc 'has teen' (có ít nhất một teen).
    -  Đơn giản hóa mô hình. Thay vì phân tích tác động của việc có 1 teen so với 2 teen, chỉ cần phân tích sự khác biệt về hành vi giữa nhóm 'có teen' và 'không có teen'. Điều này hữu ích cho các mô hình phân tích nhóm hoặc phân khúc khách hàng.

- Điền giá trị thiếu (NA) bằng 0:
    - Thay thế mọi giá trị NA (Not Available/Missing) trong cột Income bằng giá trị 0.
    - Xử lý dữ liệu thiếu để tránh lỗi khi tính toán hoặc xây dựng mô hình. Tuy nhiên, việc điền bằng 0 (thay vì giá trị trung bình/trung vị) ngụ ý rằng tin rằng các giá trị thiếu này đại diện cho những khách hàng thực sự không có thu nhập hoặc thông tin thu nhập của họ chưa được cập nhật.

- Thay đổi kiểu dữ liệu từ float sang integer (số nguyên): 
    - Chuyển cột Income từ kiểu số thực (float, có chứa số thập phân) sang kiểu số nguyên (int).
    - Tiết kiệm bộ nhớ và đảm bảo rằng dữ liệu được biểu diễn một cách chính xác hơn (vì thu nhập thường được báo cáo dưới dạng số nguyên, không có phần lẻ), giúp quá trình tính toán sau này đơn giản và hiệu quả hơn.


### Xử lý dữ liệu thiếu

```python
# Check for missing values using the isnull and sum methods
marketing_data.isnull().sum()
# Drop missing values using the dropna method
marketing_data_withoutna = marketing_data.dropna(how = 'any')
marketing_data_withoutna.shape
```

**Ý nghĩa công việc**
- Đếm tổng số giá trị thiếu theo từng cột:
    - Lệnh này thực hiện hai việc:<ul><li>isnull(): Tạo ra một DataFrame Boolean (True/False), trong đó True đại diện cho giá trị thiếu.</li><li>.sum(): Tính tổng số lượng True (giá trị thiếu) trong mỗi cột.</li></ul>
    - Giúp nhanh chóng xác định cột nào đang bị thiếu dữ liệu và mức độ thiếu (bao nhiêu giá trị thiếu), từ đó quyết định phương pháp xử lý thích hợp (ví dụ: điền giá trị, xóa hàng, hoặc xóa cột).

- Xóa tất cả các hàng có bất kỳ giá trị thiếu nào:
    - Lệnh dropna() với tham số how = 'any' sẽ loại bỏ một hàng khỏi DataFrame nếu hàng đó chứa ít nhất một giá trị thiếu ở bất kỳ cột nào. Kết quả được lưu vào DataFrame mới (marketing_data_withoutna).
    - Tạo ra một tập dữ liệu "sạch" hoàn toàn, nơi mọi quan sát (hàng) đều đầy đủ thông tin. Đây là cách tiếp cận đơn giản và hiệu quả khi số lượng giá trị thiếu ít, nhưng có thể dẫn đến mất mát dữ liệu đáng kể nếu có quá nhiều giá trị thiếu phân tán.

- Kiểm tra kích thước của DataFrame sau khi xóa: 
    - In ra số hàng và số cột của tập dữ liệu mới (marketing_data_withoutna).
    -  Xác định số lượng hàng bị mất sau khi xóa giá trị thiếu. Bằng cách so sánh số hàng này với số hàng ban đầu của marketing_data, có thể đánh giá mức độ ảnh hưởng của việc xóa dữ liệu đến tập dữ liệu của mình.
