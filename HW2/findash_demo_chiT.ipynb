```python
import requests
import bs4 as bs
import pandas as pd

def get_sp500_tickers():
    headers = {'User-agent': 'Mozilla/5.0'}
    resp = requests.get('http://en.wikipedia.org/wiki/List_of_S%26P_500_companies', headers=headers)

    soup = bs.BeautifulSoup(resp.text, 'lxml')
    table = soup.find('table', {'class': 'wikitable sortable'})

    if table is None:
        print("Lỗi: Không tìm thấy bảng S&P 500.")
        return []

    tickers = []
    # Tiếp tục cào dữ liệu...
    # LƯU Ý: Đã thay thế findAll bằng find_all (theo cảnh báo) và thêm kiểm tra lỗi
    for row in table.find_all('tr')[1:]:
        # Lấy tất cả các ô dữ liệu (td) trong hàng
        cols = row.find_all('td')

        # KIỂM TRA QUAN TRỌNG: Chỉ tiếp tục nếu hàng có ít nhất một cột (td)
        if len(cols) > 0:
            # Lấy nội dung cột đầu tiên (chứa mã cổ phiếu)
            ticker = cols[0].text.strip() # Dùng .strip() để loại bỏ ký tự xuống dòng và khoảng trắng
            tickers.append(ticker)

    # Loại bỏ các ký tự thừa (như "\n" ở cuối)
    tickers = sorted([x for x in tickers if len(x) > 0]) # Đảm bảo không thêm chuỗi rỗng
    return tickers
```

```python
def getsummary(ticker):
    # table = si.get_quote_table(ticker, dict_result = False)
    table_info = {
        "key": [],
        "value": [],
    }

    stock = yf.Ticker(ticker)

    table_info["key"].append("Previous Close")
    table_info["value"].append(stock.info["previousClose"])
    table_info["key"].append("Open")
    table_info["value"].append(stock.info["open"])
    table_info["key"].append("Bid")
    table_info["value"].append(stock.info["bid"])
    table_info["key"].append("Ask")
    table_info["value"].append(stock.info["ask"])


    table = pd.DataFrame(table_info)
    return table

# getsummary('MSFT')
```

```python
def getstockdata(ticker):
    stockdata = yf.download(ticker, period = 'max', auto_adjust=True) # Dùng auto_adjust=True cho đơn giản
    return stockdata

# getstockdata("AAPL")
```

```python
from selenium import webdriver
from selenium.webdriver.chrome.options import Options
import pandas as pd
import time

def getanalysis(ticker):
    url = f"https://finance.yahoo.com/quote/{ticker}/analysis"

    # Tùy chọn cho trình duyệt ẩn
    chrome_options = Options()
    chrome_options.add_argument("--headless")  # chạy ẩn
    chrome_options.add_argument("--no-sandbox")
    chrome_options.add_argument("--disable-dev-shm-usage")

    driver = webdriver.Chrome(options=chrome_options)
    driver.get(url)
    time.sleep(3)  # đợi trang load

    html = driver.page_source
    driver.quit()

    tables = pd.read_html(html)
    table_names = [
        "Earnings Estimate",
        "Revenue Estimate",
        "Earnings History",
        "EPS Trend",
        "EPS Revisions",
        "Growth Estimates"
    ]

    table_mapper = {name: tbl for name, tbl in zip(table_names, tables[:len(table_names)])}
    return table_mapper
```

```python
# Giả sử hàm getstockdata(ticker) đã được sửa period='max'

# 1. Lấy dữ liệu
chartdata = getstockdata(ticker)

# 2. BƯỚC SỬA LỖI QUAN TRỌNG NHẤT: Xử lý cột MultiIndex
# 2.1. Làm phẳng MultiIndex: Loại bỏ cấp thứ hai ('AAPL') của tên cột
# Các cột bây giờ sẽ là 'Close', 'High', 'Low', v.v.
chartdata.columns = chartdata.columns.droplevel(1)

# 2.2. Chuyển Index (Date) thành cột 'Date'
chartdata.reset_index(inplace=True)

# 3. Vẽ đồ thị với tên cột (chuỗi)
# Bây giờ, 'Close' là tên cột đơn giản, Plotly sẽ tìm thấy nó
fig = px.area(
    chartdata,
    x='Date',
    y='Close',
    title=f'Biểu đồ giá đóng cửa lịch sử của {ticker}'
)
```

```python
fig.update_xaxes(
            rangeselector=dict(
                buttons=list([
                    dict(count=1, label="1M", step="month", stepmode="backward"),
                    dict(count=3, label="3M", step="month", stepmode="backward"),
                    dict(count=6, label="6M", step="month", stepmode="backward"),
                    dict(count=1, label="YTD", step="year", stepmode="todate"),
                    dict(count=1, label="1Y", step="year", stepmode="backward"),
                    dict(count=3, label="3Y", step="year", stepmode="backward"),
                    dict(count=5, label="5Y", step="year", stepmode="backward"),
                    dict(label = "MAX", step="all")
                ])
            )
        )
```
